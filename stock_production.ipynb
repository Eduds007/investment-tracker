```python
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
from curl_cffi import requests
```

```python
lista =[]

with open('stocks.txt', 'r') as file:
    for line in file:
        lista.append(line.strip())
```

```python
hoje= datetime.now()
start_date = f'{hoje.year - 5}-01-01'
doze_meses = hoje - timedelta(days=365)
doze_meses= doze_meses.date()
```

```python
session = requests.Session(impersonate="chrome")
```

```python

resultado = pd.DataFrame(columns=['ticker', 'preço', 'preço-alvo', 'yield', '-1dp','dp', 'potencial', 'recomendação'])
```

```python
for stock in lista:
    print(stock)
    ação = yf.Ticker(f'{stock}.SA', session=session)

    historico = ação.history(start=start_date)
    historico['yield'] = historico['Dividends']/historico['Close']
    historico_div = historico[historico['yield'] != 0]

    historico_doze_meses = historico[historico.index >= doze_meses.__str__()]
    yield_doze_meses = historico_doze_meses['yield'].sum()

    dividends_doze_meses = historico_doze_meses['Dividends'].sum()



    agrupado = historico_div.groupby(historico_div.index.year).sum()
    try:
        if agrupado.index[-1] == hoje.year:
            agrupado.iloc[-1, -1] = yield_doze_meses
    except:
        print(agrupado)
    
    media =agrupado['yield'].mean()
    desv=agrupado['yield'].std()

    ticker = stock
    preço = historico.iloc[-1]['Close']
    target_price = dividends_doze_meses/0.06
    dyield = media
    menos_dp = media - desv
    potential = target_price/preço - 1
    
    if potential > 0 and menos_dp >= 0.06 and dyield >= 0.06 and len(agrupado) >= 5:
        recomenda = 'compra'
    elif (menos_dp >= 0.06 and dyield >= 0.06 and potential < 0 and len(agrupado) >= 5) or (dyield >= 0.06 and len(agrupado) >= 5) : 
        recomenda = 'espera'
    else:
        recomenda = 'venda'
        
     
    linha = [ticker, preço, target_price, dyield, menos_dp,desv, potential, recomenda]

    resultado = pd.concat([resultado, pd.DataFrame([linha], columns=resultado.columns)], ignore_index=True)

```
